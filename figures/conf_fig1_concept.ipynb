# Conference concept figure (Fig. 2 in the page, right of the experiment figure)

**Claim.** PCM conductance states define the weight levels of a crossbar that runs two pretrained materials models (diffusion crystal-structure generation, GNN charge-density prediction) with the same programmed weights reused over thousands of evaluations.

Source: `figures/out/fig1.png` (author's PowerPoint export; 2026-10-20 version without the conductance curve). This notebook only conditions the raster: alpha flattened on white, white margins cropped, resampled (Lanczos) to `CONCEPT_W_MM` at 600 dpi, the width it is placed at in the page (right part of the first row; the experiment figure takes the rest of the 178 mm). No panel letters (single panel).

```python
import sys, json
from pathlib import Path
sys.path.insert(0, str(Path.cwd()))
from style import *
from PIL import Image
SRC = FIG_OUT / 'fig1.png'
OUT = FIG_OUT / 'conf_fig1'
DPI = 600
PAD = 12          # px of white kept around the content (at the source resolution)
WHITE_TOL = 245   # a pixel is background if all channels >= this
CONCEPT_W_MM = 116.0   # placed width in the page row (experiment 58 mm + 4 mm gap + concept = 178 mm)
```

```python
im = Image.open(SRC)
if im.mode in ('RGBA', 'LA', 'P'):
    im = im.convert('RGBA'); bg = Image.new('RGBA', im.size, (255, 255, 255, 255)); im = Image.alpha_composite(bg, im)
im = im.convert('RGB')
a = np.asarray(im)
ys, xs = np.where((a < WHITE_TOL).any(-1))
im = im.crop((max(xs.min() - PAD, 0), max(ys.min() - PAD, 0), min(xs.max() + PAD + 1, im.width), min(ys.max() + PAD + 1, im.height)))
w_px = int(CONCEPT_W_MM * MM * DPI)
out = im.resize((w_px, round(im.height * w_px / im.width)), Image.LANCZOS)
OUT.mkdir(parents=True, exist_ok=True)
out.save(OUT / 'conf_fig1.png', dpi=(DPI, DPI))
print('source', Image.open(SRC).size, '-> cropped', im.size, '-> out', out.size,
      f'= {out.width / DPI * 25.4:.1f} x {out.height / DPI * 25.4:.1f} mm')
json.dump({'width_mm': out.width / DPI * 25.4, 'height_mm': out.height / DPI * 25.4}, open(OUT / 'size.json', 'w'))
display(NotebookImage(filename=str(OUT / 'conf_fig1.png'), width=int(out.width * 0.25)))
```
