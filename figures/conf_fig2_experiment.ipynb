# Conference experiment figure (Fig. 1 in the page, left of the concept figure) — placeholders

**Claim.** The fabricated PCM device (structure, microstructure, material phase) shows the multilevel switching whose measured conductance states drive all simulations.

- **a** Device structure (schematic / optical image)
- **b** Cross-sectional TEM
- **c** Material characterization (XRD / Raman / EDS)
- **d** Electrical switching (I–V or R–V)
- **e** Multilevel programming / retention

Drop the experimental images as `figures/exp/{a..e}.png` (already labelled, any aspect); each panel shows the image if it exists, else a grey placeholder box. Slot size: see the layout cell.

Canvas: `EXP_W_MM` = 58 mm wide (left part of the first page row), height = the concept figure's (`figures/out/conf_fig1/size.json`), so both sit on one row. Slots: a–c on top, d–e below.

```python
import sys, json
from pathlib import Path
sys.path.insert(0, str(Path.cwd()))
from style import *
from matplotlib.lines import Line2D
EXP_W_MM = 58.0     # 第一行左侧：实验图宽度（右侧概念图 116 mm，间隔 4 mm）
W = EXP_W_MM * MM
EXP = ROOT / 'figures' / 'exp'
TITLES = {'a': 'Device\nstructure', 'b': 'Cross-sect.\nTEM', 'c': 'Material\ncharact.',
          'd': 'Electrical switching\n(I–V / R–V)', 'e': 'Multilevel /\nretention'}
```

```python
# 字体设置 / fonts —— 全文唯一来源（style.PAPER_FONTS）
FS = use_paper_fonts()
FS_TITLE, FS_LABEL, FS_TICK = FS['title'], FS['label'], FS['tick']
FS_LEGEND, FS_ANNOT, FS_PANEL = FS['legend'], FS['annot'], FS['panel']
FS_CBLABEL, FS_CBTICK = FS['cbar_label'], FS['cbar_tick']
```

```python
# layout: two rows, a–c on top, d–e below; canvas EXP_W_MM x height of the concept figure
# Derivation (inches): left 0.12 (letters), gaps 0.14 (letters), right 0.02, top 0.12 (letters), row gap 0.13, bottom 0.02
#   top slot w = (W - 0.12 - 2*0.14 - 0.02) / 3; bottom slot w = (W - 0.12 - 0.14 - 0.02) / 2; slot h = (H - 0.27) / 2
OUT = FIG_OUT / 'conf_fig2'
PANEL_OUT = OUT / 'panels'
PREVIEW_SCALE = 0.5
H = json.loads((FIG_OUT / 'conf_fig1' / 'size.json').read_text())['height_mm'] * MM
L0, GX, R0, T0, GY, B0 = 0.12, 0.14, 0.02, 0.12, 0.13, 0.02

def make_layout():
    fig = plt.figure(figsize=(W, H))
    sh = (H - T0 - GY - B0) / 2
    w3 = (W - L0 - 2 * GX - R0) / 3; w2 = (W - L0 - GX - R0) / 2
    top = [(L0 + k * (w3 + GX), B0 + sh + GY, w3, sh) for k in range(3)]
    bot = [(L0 + k * (w2 + GX), B0, w2, sh) for k in range(2)]
    return fig, [fig.add_axes([x / W, y / H, w / W, h / H]) for x, y, w, h in top + bot]

def panel_run(i, letter, draw):
    fig, slots = make_layout()
    for k, ax in enumerate(slots):
        if k != 'abcde'.index(letter): ax.remove()
    draw(fig, slots['abcde'.index(letter)])
    group = {'axes': list(fig.axes), 'texts': list(fig.texts)}
    boxes = save_matched_panels(fig, {letter: group}, PANEL_OUT); plt.close(fig)
    display(NotebookImage(filename=str(PANEL_OUT / f'{letter}.png'), width=int(boxes[letter].width * 600 * PREVIEW_SCALE)))
```

```python
# ---------------- 本面板配色（a–e 共用，占位框） ----------------
Cx_FILL = '#F2F2F2'   # 占位框填充
Cx_EDGE = '#9E9E9E'   # 占位框虚线边框
Cx_TXT = '#7E7E7E'    # 占位文字
# ----------------------------------------------------------------------
def draw_slot(fig, ax, letter):
    img = EXP / f'{letter}.png'
    if img.exists():
        ax.imshow(plt.imread(img)); ax.axis('off'); return
    ax.set_facecolor(Cx_FILL)
    for sp in ax.spines.values():
        sp.set_visible(True); sp.set_linestyle((0, (3, 2))); sp.set_color(Cx_EDGE); sp.set_linewidth(0.5)
    ax.set_xticks([]); ax.set_yticks([])
    ax.text(0.5, 0.5, TITLES[letter] + '\n(experiment)', ha='center', va='center', transform=ax.transAxes,
            fontsize=FS_ANNOT - 1, color=Cx_TXT)

for L in 'abcde':
    globals()['draw_' + L] = (lambda L: (lambda fig, ax: draw_slot(fig, ax, L)))(L)
    panel_run(0, L, globals()['draw_' + L])
```

```python
# composite
fig, slots = make_layout()
for L, ax in zip('abcde', slots):
    globals()['draw_' + L](fig, ax)
    mark(fig, ax, L, dx_in=-0.12, dy_in=0.0)
export_composite(fig, OUT, 'conf_fig2')
plt.close(fig)
display(NotebookImage(filename=str(OUT / 'conf_fig2.png'), width=int(W * 600 * PREVIEW_SCALE)))
```
